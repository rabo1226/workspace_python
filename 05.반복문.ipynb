## 반복문

```python
#일반 반복문

#1~10 출력
for e in range(1, 11) :
  print(e)
```

```python
#10~30까지의 정수 중 짝수만 출력
for e in range(10, 31) : 
  if e % 2 == 0 :
    print(e)
    
```

```python
# 파이썬은 변수명 사용 시 주의 필요!
# 파이썬 내장 함수명과 동일한 이름의 변수를 사용하면 내장함수가 작동하지 않음!!!!
# 대표적 함수 : sum(), max(), min(), len(), list(), dic()
# 실수로 내장된 함수와 동일한 이름으로 변수를 만들었다면 해당 변수는 삭제해야한다!

#1~10까지의 합을 출력
#이 문제를 풀기 위해 sum변수 생성! -> 문제가 발생할 수 있는 코드
sum_result = 0
for e in range(1, 11) : 
  sum_result = sum_result + e
print(sum_result)
```

```python
#sum변수 사용에는 문제가 없지만, 내장된 sum() 함수가 호출되지 않는 문제 발생!
#print(sum)
print(sum_result)

list1 = [1, 2, 3]

print(sum(list1))
```

```python
# for-each

list1 = [1, 2, 3, 4, 5]
for e in list1 :
  print(e)

print('\n')
for e in [1, 2, 3] :
  print(e)
```

```python
#반복 index 데이터
#반복 index를 사용할 수 있는 for문
# for index변수, 데이터 하나하나를 의미하는 변수 in enumerate(리스트변수) : 실행문
list1 = [2, 3, 4, 6, 8, 10]
for a, b in enumerate(list1) : 
  print(f'a = {a}, b = {b}')
```
