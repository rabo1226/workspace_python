## 딕셔너리(dictionary)
자바스크립트의 객체 자료형과 동일하다.  
key와 value를 한 쌍으로 다수의 데이터를 가질 수 있는 자료형

```python
#빈 딕셔너리 데이터 생성
dic1 = {}

#데이터있는 딕셔너리 생성
dic2 = {
  'name' : 'kim',
  'age' : 20
}

#데이터 출력
print(dic2)  #전체 출력

print(dic2['name'])  #데이터 선택 출력   -> 딕셔너리[key]
print(dic2['age'])  #데이터 선택 출력
```

```python
#딕셔너리 데이터 추가 및 수정

phone = {
  'madelName' : 's20',
  'price' : 10000
}
print(phone)

#데이터 추가 : 딕셔너리가 가지지 않은 key를 사용하면 데이터 추가
phone['color'] = 'red'
print(phone)

#print(phone['brand']) #존재하지 않는 key값의 데이터를 읽는 것은 불가능

#데이터 수정 : 기존 key에 데이터를 저장하면 수정
phone['price'] = 20000
print(phone)

#데이터 삭제
del phone['madelName']
#del phone   : 딕셔너리 phone 자체를 삭제.
print(phone)

```
