```python
# Pandas 라이브러리를 사용하기 위해서 import를 우선 진행
# pandas 라이브러리를 import 하고, 사용 시 pd라는 이름 as pd (별칭 pd)으로 사용하겠다는 명령어
import pandas as pd
```

```python
# Pandas 사용을 위한 딕셔너리 데이터
data = {
  '이름' : ['김자바', '이자바', '박자바', '최캐드', '정캐드'],
  '학급' : ['자바반', '자바반', '자바반', '캐드반', '캐드반'],
  '국어점수' : [90, 80, 70, 100, 90],
  '영어점수' : [70, 80, 80, 70, 90],
  '평균' : [80.0, 80.0, 75.0, 85.0, 90.0],
  '취미' : ['독서', '', '', '요가', '축구']
}

#딕셔너리 데이터를 DataFrame(표 형태)으로 변환
df = pd.DataFrame(data)
```

```python
#데이터 확인 / print() / df (데이터자체)
df
```

```python
#df 데이터를 엑셀 파일로 저장 .xlsx
# 행 번호도 컬럼 데이터로 저장된다.
# 엑셀 파일로 변환 시 행번호 제외시킬 수 있다. to_excel 함수의 두번째 매개변수로 index=False
df.to_excel('test2.xlsx', index = False)
```

```python
#저장된 엑셀 파일 불러오기 .xlsx
new_df = pd.read_excel('test2.xlsx')
new_df
```
