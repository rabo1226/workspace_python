```python
#padas 설치는 최초 1회만! 그러나 import는 매번 실행 필요!
import pandas as pd
```

```python
df = pd.read_excel('data.xlsx')
df
```

```python
#특정 컬럼 데이터만 조회하기
df[['국어']]

#이름, 국어 컬럼  (컬럼 조회 시 컬럼명을 리스트 형태로 설정해야한다!)
df[['이름', '국어']]
```

```python
#특정 열과 행 데이터 조회
#1. df.loc[행 번호, 열 이름] / 이름으로 조회하겠다!
#2. df.iloc[행 번호, 열 번호(index)] / 순번으로 조회하겠당!

df.loc[2]       # 2번행 전체 열 데이터 조회
df.loc[1 : 4]   # 1~4번행 열 데이터 조회
df.loc[1 : ]    # 1행부터 전체 열 데이터 조회

df.loc[0, '이름']   # 0번행 '이름' 데이터 조회
df.loc[0, ['이름']]
df.loc[1 : 4, ['이름', '성별']] # 1~4번행, [이름, 성별] 데이터 조회

df.iloc[1 : 3, [0, 2]]  # 1~3번행, [0, 2]번 컬럼 데이터 조회
df.iloc[ : , 1 : 5]
```

```python
# 특정 조건에 만족하는 데이터 조회
# 국어 점수가 80점 이상인 데이터만 조회

#1. 국어점수가 80점 이상이라는 조건
df['국어'] >= 80

#2. 만든 조건을 적용하는 코드 (전체 컬럼 조회)
df[ df['국어'] >= 80 ]

# 위와 같은 코드를 loc를 사용 (컬럼의 조건도 추가 가능!)
df.loc[ df['국어'] >= 80 ]   #국어를 기준으로 행을 조회하는 것! =  조건을 행 조건에 넣어야한당!

# 성별이 '여'인 데이터만 조회
df[ df['성별'] == '여']
df.loc[df['성별'] == '여', ['이름', '성별']]
df.iloc[df['성별'] == '여' , 0 : 3]

# 국어가 80점 이상이면서 성별은 여인 학생만 조회
conditon1 = df['국어'] >= 80 
conditon2 = df['성별'] == '여'

df[conditon1 & conditon2] # and
df[conditon1 | conditon2] # or
# 여러개의 조건을 직접 작성하려면 조건마다 무조건 소괄호에 감싸야한다.
df[(df['국어'] >= 80) & (df['성별'] == '여')]
df.loc[conditon1 & conditon2]


# 팁)  조건에 맞는 데이터 조회 시 query() 함수를 사용할 수 있음
#df.query('국어 >= 80')
#df.query(' 이름 == "구이서" ')
df.query(' 국어 >= 80 and 성별 == "여" ')
```
