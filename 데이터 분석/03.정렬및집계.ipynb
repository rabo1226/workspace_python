#### 데이터 정렬
##### df.sort_values(by = [정렬기준이 되는 컬럼], ascending = true)

```python
import pandas as pd
```

```python
df = pd.read_excel("data.xlsx")
df
```

```python
# '이름' 컬럼 기준으로 오름차순 정렬
df.sort_values(by=['이름'] , ascending=True)
# 조회 컬럼[], 정렬 기준[] 가능! 
df.sort_values(by=['이름'] , ascending=[True])
```

```python
#'반' 기준 내림차순 정렬 후 이름 기준 오름차순 정렬
df.sort_values(by=['반', '이름'], ascending=[False, True])
```

```python
#데이터 집계하기
```

```python
#전체적인 집계 통계 확인
df.describe()
```

```python
#'반' 컬럼의 카테고리(a,b반)별 데이터 확인
df['반'].value_counts()
```

```python
#특정 컬럼의 집계 정보 확인

#'국어'점수의 합
#df['국어'].sum() -> 합이 나오긴 하지만 숫자형태가 아님

#집계데이터를 숫자형태로 보기위한 방법
#print(df['국어'].sum()) #출력문 사용
float(df['국어'].sum())     #데이터를 실수형으로 변환 = 데이터 저장 가능, 정수 변환 : int() / 실수 변환 : float()
float(df['국어'].mean())    #평균
float(df['국어'].max())     #최댓값
float(df['국어'].min())     #최소값
float(df['국어'].count())   #갯수
float(df['국어'].median())  #중앙값

#여러 컬럼의 통계를 한 번에 확인할 수 있음
df[['국어', '수학']].mean()
```

```python
#그룹별 집계 정보
#'반'별 국어점수의 평균
df.groupby('반')['국어'].mean()

#'성별' 영어점수의 최고값
df.groupby('성별')['수학'].max()

#'성별' 인원수
df.groupby('성별')['이름'].count()
df.groupby('성별')['수학'].count()   #두 집계데이터의 차이 : NaN(결측치 Not A Number 값) / NA(Not Available)여부로 달라짐!

#'반','성별' 국어점수의 평균
df.groupby(['반','성별'])['국어'].mean()

#여러 컬럼 선택
#'반'별 국어, 수학점수의 평균
df.groupby(['반'])[['국어', '수학']].mean()

#여러 집계합수 사용
#'반'별 수학점수의 평균, 최고값, 최저값, 중앙값
# agg -> Aggregation : 집합
df.groupby('반')['수학'].agg(['mean', 'max', 'min', 'median', 'count'])
```
