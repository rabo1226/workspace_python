```python
#1. 1부터 키보드로 입력받은 정수까지의 합 출력
num = int(input('정수입력 : '))
sum_result = 0
for e in range(1, num + 1) : 
  sum_result = sum_result + e
print(sum_result)
```

```python

#2. 키보드로 입력받은 단수의 구구단 출력
# ex) 키보드 3 입력 3 * 1 = 3  ~~~~ 3 * 9 = 27

key_num = int(input('입력 단 : '))
print('구구단')
for e in range(1, 10) :
  e1 = int(e)
  print(f'{key_num} * {e} = {key_num * e}')
```

```python
#3. 아래 제시된 리스트에서 가장 큰 요소와 가작 작은 요소의 차를 출력
sample_list = [3, 1, 7, 10, 5]

print(max(sample_list) - min(sample_list))
```

```python
#4. 아래 제시된 두 리스트의 모든 요소의 합 출력
list1 = [1, 3, 5]
list2 = [10, 15, 20]

list3 = list1 + list2
sum_num1 = sum(list3)
print(sum_num1)
```

```python
#5. 아래 데이터에서 't'의 갯수를 출력
# hint : 파이선은 문자열도 리스트로 취급
# name = 'hong' -> name[1] = 'o'
data = ['mother', 'father', 'brother']
tcnt = 0

for word in data :
  for c in word :
    if c == 't' : 
      tcnt = tcnt + 1

print(tcnt)
```
