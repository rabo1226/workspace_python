```python
#파이썬은 함수 정의 시 매개변수에 default 값을 줄 수 있다.   함수의 호출에서 매개변수 값 저장과는 다른 의미!!!!!!
def tellMyIntro(name, age = 100) :                       #매개변수 전달이 없다면 age에는 100값이 기본값으로 설정된당!
  print(f'내 이름은 {name}, 나이는 {age}살 입니다.')      
```

```python
tellMyIntro('Kim', 50)
tellMyIntro('Lee')
tellMyIntro(age = 90, name = 'Park')
```

```python
print(1, end = '!')    #기본 매개변수 외에는 기본값이 저장되어있다! \n이 개행이 기본값!!
print(2)
print(20, end = '')
print('오잉', end = '20')
print()
```

```python
#2. 첫번째 매개변수로 전달되는 딕셔너리 데이터에 두번째 매개변수로 전달된 key와 
#   세번째 전달된 매개변수로 전달된 vlaue값을 추가한 딕셔너리 데이터를 리턴하는 함수 선언 및 호출

def test2(dic_data, key, value) : 
  dic_data[key] = value
  return dic_data

person = {
  'name' : 'kim',
  'age' : 20,
}
testResult = test2(person, 'addr', '울산시')
print(testResult)
```
