## 리스트
파이썬의 리스트는 자바스크립트의 배열과 기본적이 동작이 같다.  
다만, 파이썬에서는 리스트의 내용을 컨트롤하는 다양한 기능이 있다.

```python
#빈 리스트 생성
list1 = []

#데이터가 있는 리스트 생성
list2 = [1, 2, 3.5, '문자열', {}, []]

print(list1)
print(list2)

print(list2[1])
```

```python
list3 = [1, 2, 3]

#리스트 데이터 추가
# 1. append() : list의 마지막 요소에 데이터 추가
list3.append(4)
print(list3)

# 2. insert(index, 데이터) : list의 선택 index에 데이터 추가
list3.insert(2, 5)   #리스트의 2번째 요소에 5 추가
print(list3)
```

```python
#리스트 데이터 삭제
list5 = [1, 2, 3, 4, 5]

#1번 요소 데이터 삭제
del list5[1]
print(list5)

del list5[2]
print(list5)

#del list5 #list5 변수 자체를 제거
#print(list5)  #=> list5자체를 제거했기 때문에 오류발생!
```

```python
#1. 파이썬에만 있는 문법 
list1 = [1, 2, 3]
list2 = [4, 5, 6]

#위 두 리스트를 나열(합하기)
list3 = list1 + list2
print(list3)

list4 = [1, 2, 3] #[1, 2, 3, 1, 2, 3]
list5 = list4 * 3
print(list5)
```

```python
#2. 파이썬에만 있는 문법 - 슬라이싱 연산(리스트에서 일부 데이터만 추출)
#list[시작index: 시작index-1]
list1 = [1, 2, 3,  4, 5]   #학원 시험 문제!

print(list1[2:4]) #2번재 요소부터 3번째 요소까지
print(list1[1:]) #1번째 요소부터 마지막 요소까지
print(list1[:3]) #0번째 요소부터 2번째 요소까지
print(list1[:]) #모든 데이터
print(list1[1:-2])  #1번째 요소부터 -2(-1)번재 요소(-5, -4, -3, -2, -1) => 1번째 요소부터 -2초과의 요소까지
```

```python
list1 = [1, 2, 3, 4, 5]

#리스트의 길이
print(len(list1))


#리스트에서 가장 큰 값
print(f'가장 큰 값 :', max(list1))
print(f'가장 큰 값 :', min(list1))
print(sum(list1))
```
